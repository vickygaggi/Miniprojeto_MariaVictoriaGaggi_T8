### 1.1. Carga Automatizada e Importação das Bibliotecas
Para garantir a portabilidade e a execução contínua do projeto em diferentes sessões do Google Colab, foi desenvolvida a função `carregar_base_varejo()`.

A função possui parâmetro e retorno, verificando se o arquivo `Base Varejo.csv` já está presente no ambiente local de execução. Caso não esteja, ela aciona automaticamente a interface de upload do Colab (`files.upload()`) e realiza a leitura utilizando o delimitador ponto e vírgula (`;`), conforme a estrutura original do arquivo.

```python
# Carregando arquivo google colab
import os
import pandas as pd
from google.colab import files

# Função para carregar a base de dados do varejo ***
def carregar_base_varejo(caminho_arquivo = 'Base Varejo.csv'):

  # Verifica se o arquivo já está presente no ambiente
  if not os.path.exists(caminho_arquivo):
    print(f'O arquivo {caminho_arquivo} não foi encontrado automaticamente. Por favor, faça o upload do arquivo.')

    # Carrega o arquivo
    uploaded = files.upload()

    print(f'O arquivo {caminho_arquivo} foi carregado com sucesso!')
  else:
    print(f'O arquivo {caminho_arquivo} já está presente no ambiente')

  # Realiza a leitura do arquivo CSV
  df = pd.read_csv(caminho_arquivo, sep = ';')

  # Retorna o DataFrame carregado
  return df
```

```python
# --- Executando a função de carga ---
df = carregar_base_varejo()
```

### 1.2. Inspeção Estrutural (Linhas, Colunas e Amostra)
Nesta subetapa, inspecionamos o volume total de dados brutos e os primeiros registros da base para compreender a granularidade das informações registradas (onde cada linha representa um item adquirido em uma compra realizada por determinado cliente).

### 1.4. Diagnóstico de valores ausentes (nulos reais - `NaN`)
Quantidade e percentual de valores nulos em cada coluna. Também identificamos as colunas totalmente vazias.

```python
nulos = df.isnull().sum()
tabela_nulos = pd.DataFrame({'nulos': nulos, '% nulos': (nulos / len(df) * 100).round(2)})
display(tabela_nulos)

# Colunas onde todos os valores são nulos
colunas_vazias = df.columns[df.isnull().all()].tolist()
print(f'Colunas 100% vazias ({len(colunas_vazias)}): {colunas_vazias}')

# Nulos nas demais colunas
nulos_restantes = df.drop(columns=colunas_vazias).isnull().sum().sum()
print(f'Nulos nas demais colunas: {nulos_restantes}')
```

### 1.5. Diagnóstico de Linhas Duplicadas
Identificamos a quantidade e o percentual de registros totalmente duplicados na base. Compras diferentes efetuadas pelo mesmo cliente não são consideradas duplicadas; porém, linhas em que todas as variáveis são exatamente idênticas precisam ser contabilizadas. Como a base não possui coluna de quantidade, uma linha repetida pode representar o mesmo produto comprado mais de uma vez na mesma compra, e por isso a decisão de removê-las será tomada na Sprint 2.

```python
qtd_duplicadas = df.duplicated().sum()
pct_duplicadas = qtd_duplicadas / len(df) * 100
print(f'Quantidade de linhas duplicadas: {qtd_duplicadas:,}'.replace(',', '.'))
print(f'Percentual de duplicadas: {pct_duplicadas:.2f}%'.replace('.', ','))
```

### 1.6. Diagnóstico de Textos Vazios ou Preenchidos com Espaços
Percorremos todas as colunas de texto para verificar se existem valores vazios, formados apenas por espaços ou com espaços nas extremidades (`.str.strip()`). Registramos as quantidades encontradas, inclusive quando são zero.

```python
colunas_texto = df.select_dtypes(exclude='number').columns

for col in colunas_texto:
    vazios = (df[col].str.strip() == '').sum()
    com_espacos = (df[col] != df[col].str.strip()).sum()
    print(f'{col}: vazios ou só espaços = {vazios} | espaços nas pontas = {com_espacos}')
```

### 1.7. Validação de Inconsistências e Valores Únicos em Colunas Textuais
Para identificar inconsistências textuais, contamos os valores únicos de cada coluna de texto e mostramos a distribuição completa das colunas com poucas categorias. Também verificamos variações de maiúsculas e minúsculas e buscamos o valor `#N/D`, que representa um dado ausente escrito como texto e que o `isnull()` não detecta.

```python
colunas_texto = df.select_dtypes(exclude='number').columns

for col in colunas_texto:
  qtd_unicos = df[col].nunique()
  print(f"\nColuna: {col} -> {qtd_unicos} valores únicos")

  # Se tem poucas categorias, mostramos todas com a contagem
  if qtd_unicos <= 10:
    print(df[col].value_counts().to_string())

  # Variações de maiúsculas/minúsculas (ex.: 'Bebidas' e 'BEBIDAS' contariam como 2)
  variacoes = qtd_unicos - df[col].str.strip().str.upper().nunique()
  print(f"Variações de maiúsculas/minúsculas: {variacoes}")

  # Procura o valor '#N/D' (dado ausente escrito como texto)
  qtd_nd = (df[col] == '#N/D').sum()
  print(f"Ocorrências de '#N/D': {qtd_nd}")

# Verifica a quais produtos pertencem as linhas com '#N/D'
print("\nPR_ID das linhas com '#N/D' em PR_CAT:")
print(df[df['PR_CAT'] == '#N/D']['PR_ID'].value_counts().to_string())
```
