### 1.1. Carga Automatizada e Importação das Bibliotecas
Para garantir a portabilidade e a execução contínua do projeto em diferentes sessões do Google Colab, foi desenvolvida a função `carregar_base_varejo()`.

A função possui parâmetro e retorno, verificando se o arquivo `Base Varejo.csv` já está presente no ambiente local de execução. Caso não esteja, ela aciona automaticamente a interface de upload do Colab (`files.upload()`) e realiza a leitura utilizando o delimitador ponto e vírgula (`;`), conforme a estrutura original do arquivo.

```python
# Carregando arquivo google colab
import os
import pandas as pd
from google.colab import files

# Função para carregar a base de dados do varejo ***
def carregar_base_varejo(caminho_arquivo = 'Base Varejo.csv'):

  # Verifica se o arquivo já está presente no ambiente
  if not os.path.exists(caminho_arquivo):
    print(f'O arquivo {caminho_arquivo} não foi encontrado automaticamente. Por favor, faça o upload do arquivo.')

    # Carrega o arquivo
    uploaded = files.upload()

    print(f'O arquivo {caminho_arquivo} foi carregado com sucesso!')
  else:
    print(f'O arquivo {caminho_arquivo} já está presente no ambiente')

  # Realiza a leitura do arquivo CSV
  df = pd.read_csv(caminho_arquivo, sep = ';')

  # Retorna o DataFrame carregado
  return df
```

```python
# --- Executando a função de carga ---
df = carregar_base_varejo()
```

### 1.4. Diagnóstico de Valores Ausentes (Nulos)
Mapeamos a quantidade de valores nulos (`NaN`) presentes em cada uma das colunas da base de dados bruta.
*(Nota: O registro das colunas nulas identificadas aqui servirá de insumo para o tratamento e remoção de colunas fantasmas na Sprint 2).*

```python
valores_nulos = df.isnull().sum()
print(valores_nulos)
```

### 1.5. Diagnóstico de Linhas Duplicadas
Identificamos a quantidade exata de registros totalmente duplicados na base. Respeitando as regras de negócio do projeto, lembramos que compras diferentes efetuadas pelo mesmo cliente não são consideradas duplicadas; contudo, registros em que todas as variáveis são exatamente idênticas caracterizam duplicidade de lançamento e precisam ser contabilizados.

```python
qtd_duplicadas = df.duplicated().sum()
print(f'Quantidade de duplicadas: {qtd_duplicadas}')
```

### 1.6. Diagnóstico de Textos Vazios ou Preenchidos com Espaços
Varremos todas as colunas do tipo textual (`object`) utilizando a remoção de espaços em branco nas extremidades (`.str.strip()`), garantindo que não existam registros contendo apenas strings vazias (`''`), o que poderia mascarar a ausência de dados válidos.

```python
colunas_texto = df.select_dtypes(include=['object']).columns
for col in colunas_texto:
  vazios = df[col].astype(str).str.strip().eq('').sum()
  print(f'Coluna: {col} - Textos vazios ou espaços: {vazios}')
```

### 1.7. Validação de Inconsistências e Valores Únicos em Colunas Textuais
Para identificar possíveis inconsistências textuais (como variações de maiúsculas e minúsculas, erros de digitação ou espaços ocultos), inspecionamos os valores únicos presentes nas colunas categóricas da base.

```python
colunas_texto = df.select_dtypes(include=['object']).columns
for col in colunas_texto:
  unicos = df[col].unique()
  print(f"\nColuna: '{col}'")
  print(f"Quantidade de valores únicos: {len(unicos)}")
  print(f"Alguns valores encontrados: {unicos[:10]}")
```
