### 1.1. Carga Automatizada e Importação das Bibliotecas
Para garantir a portabilidade e a execução contínua do projeto em diferentes sessões do Google Colab, foi desenvolvida a função `carregar_base_varejo()`.

A função possui parâmetro e retorno, verificando se o arquivo `Base Varejo.csv` já está presente no ambiente local de execução. Caso não esteja, ela aciona automaticamente a interface de upload do Colab (`files.upload()`) e realiza a leitura utilizando o delimitador ponto e vírgula (`;`), conforme a estrutura original do arquivo.

```python
# Carregando arquivo google colab
import os
import pandas as pd
from google.colab import files

# Função para carregar a base de dados do varejo ***
def carregar_base_varejo(caminho_arquivo = 'Base Varejo.csv'):

  # Verifica se o arquivo já está presente no ambiente
  if not os.path.exists(caminho_arquivo):
    print(f'O arquivo {caminho_arquivo} não foi encontrado automaticamente. Por favor, faça o upload do arquivo.')

    # Carrega o arquivo
    uploaded = files.upload()

    print(f'O arquivo {caminho_arquivo} foi carregado com sucesso!')
  else:
    print(f'O arquivo {caminho_arquivo} já está presente no ambiente')

  # Realiza a leitura do arquivo CSV
  df = pd.read_csv(caminho_arquivo, sep = ';')

  # Retorna o DataFrame carregado
  return df
```

```python
# --- Executando a função de carga ---
df = carregar_base_varejo()
```

# Sprint 2: Limpeza e Padronização

### 2.1. Cópia de trabalho

# Sprint 3: Funções e Organização

Nesta etapa transformamos o tratamento feito na Sprint 2 em funções reutilizáveis, com parâmetros e retorno. Assim, o mesmo processo pode ser aplicado a qualquer base com a mesma estrutura, e o projeto segue uma sequência clara: **leitura → limpeza → análise**.

| Função | Parâmetro | Retorno |
|---|---|---|
| `carregar_base_varejo()` | caminho do arquivo CSV | DataFrame bruto (criada na Sprint 1) |
| `tratar_dados()` | DataFrame bruto e lista de colunas de texto | DataFrame tratado |
| `gerar_resumo()` | DataFrame | Tabela-resumo com os principais indicadores |

### 3.1. Função de tratamento: `tratar_dados()`
A função recebe o DataFrame bruto e aplica as mesmas decisões da Sprint 2, na mesma ordem: remove colunas vazias, padroniza o texto, preenche `#N/D` com `NAO INFORMADO`, remove duplicadas idênticas e converte `DATA` para o tipo data. Ela trabalha sobre uma **cópia** (`.copy()`), por isso o DataFrame original não é alterado, e devolve o DataFrame tratado.

```python
def tratar_dados(df, colunas_texto = ['CL_GENERO', 'CL_SEG', 'PR_CAT', 'PR_NOME']):

  # Trabalha em uma cópia para não alterar o DataFrame original
  df_tratado = df.copy()

  # 1. Remove colunas totalmente vazias
  df_tratado = df_tratado.dropna(axis=1, how='all')

  # 2. Padroniza o texto (remove espaços nas pontas e converte para maiúsculas)
  for col in colunas_texto:
    df_tratado[col] = df_tratado[col].str.strip().str.upper()

  # 3. Preenche '#N/D' com 'NAO INFORMADO'
  df_tratado[colunas_texto] = df_tratado[colunas_texto].replace('#N/D', 'NAO INFORMADO')

  # 4. Remove linhas duplicadas
  df_tratado = df_tratado.drop_duplicates().reset_index(drop=True)

  # 5. Converte DATA para o tipo data
  df_tratado['DATA'] = pd.to_datetime(df_tratado['DATA'], format='%d/%m/%Y')


  # Retorna o DataFrame tratado
  return df_tratado
```

### 3.2. Função de resumo: `gerar_resumo()`
A função recebe um DataFrame (bruto ou tratado) e devolve uma tabela com os principais indicadores. Ela separa o que são **registros** (linhas, ou seja, itens de compra), **compras** (`CO_ID` distintos), **clientes** (`CL_ID` distintos) e **produtos** (`PR_ID` distintos), para evitar confundir essas medidas nas próximas etapas.

```python
def gerar_resumo(df):

  # Converte DATA para data apenas para calcular o período (funciona com a base bruta e a tratada)
  data = pd.to_datetime(df['DATA'], format='%d/%m/%Y')

  # Linhas em que PR_CAT ou PR_NOME têm o valor (cada linha é contada uma vez)
  linhas_nd = ((df['PR_CAT'] == '#N/D') | (df['PR_NOME'] == '#N/D')).sum()
  linhas_nao_informado = ((df['PR_CAT'] == 'NAO INFORMADO') | (df['PR_NOME'] == 'NAO INFORMADO')).sum()

  resumo = {
      'Registros (linhas)': len(df),
      'Colunas': df.shape[1],
      'Valores nulos (NaN)': df.isnull().sum().sum(),
      'Linhas duplicadas': df.duplicated().sum(),
      'Linhas com #N/D': linhas_nd,
      'Linhas com NAO INFORMADO': linhas_nao_informado,
      'Compras distintas (CO_ID)': df['CO_ID'].nunique(),
      'Clientes distintos (CL_ID)': df['CL_ID'].nunique(),
      'Produtos distintos (PR_ID)': df['PR_ID'].nunique(),
      'Primeira data': data.min().date(),
      'Última data': data.max().date()
      }

  # Retorna o resumo como tabela
  return pd.DataFrame.from_dict(resumo, orient='index', columns=['Valor'])
```

### 3.3. Execução em sequência: leitura, limpeza e resumo
Executamos o fluxo completo a partir da base bruta: carregar, tratar e resumir. Também conferimos que a função `tratar_dados()` produz exatamente o mesmo resultado obtido passo a passo na Sprint 2.

```python
# leitura
df_bruto = carregar_base_varejo()

# limpeza
df_tratado = tratar_dados(df_bruto)

# Resumo antes e depois
resumo_antes = gerar_resumo(df_bruto)
resumo_depois = gerar_resumo(df_tratado)

comparacao_resumo = pd.concat([resumo_antes, resumo_depois], axis=1)
comparacao_resumo.columns = ['Antes', 'Depois']

print(comparacao_resumo)

# Conferência: a função deve gerar o mesmo resultado da Sprint 2
print('\nMesmo resultado da Sprint 2:', df_tratado.equals(df_limpo))
```

**Conclusão - Sprint 3: Funções e Organização**

*Funções criadas:* `tratar_dados(df, colunas_texto)` recebe a base bruta e devolve a base tratada; `gerar_resumo(df)` recebe uma base e devolve uma tabela de indicadores. Ambas possuem parâmetros e retorno e foram chamadas no fluxo.

*Reprodutibilidade:* A função de tratamento gerou o mesmo resultado obtido passo a passo na Sprint 2 (733.447 registros e 10 colunas), o que permite repetir a limpeza a partir da base original.

*Efeito da limpeza:* Os registros passaram de 830.000 para 733.447, mas o número de compras (18.471), clientes (1.000) e produtos (229) permaneceu igual, indicando que as duplicadas eram linhas repetidas dentro de compras existentes.

# Sprint 4: Análise

Nesta etapa aplicamos um filtro à base tratada, calculamos estatísticas básicas de uma coluna numérica e produzimos dois agrupamentos. Como cada linha da base representa **um item de uma compra**, é importante não confundir as medidas:

| Medida | O que representa | Como é calculada |

A base **não possui coluna de valor ou de quantidade**, portanto não é possível calcular faturamento; as análises usam contagens. O significado das colunas segue o dicionário de dados da base.

| Decisão | Escolha | Justificativa |
|---|---|---|
| Filtro | Registros de 2021 | Ano com o maior número de registros da base |
| Coluna numérica | `CL_FHL` (número de filhos) | Medida numérica com significado de negócio |
| Unidade de análise | Cliente | `CL_FHL` é uma característica do cliente, repetida em cada item comprado |
| Agrupamento 1 | Registros por categoria (`PR_CAT`) | Mede itens vendidos por categoria |

### 4.1. Filtro: registros de 2021

### 4.2. Estatísticas básicas: número de filhos (`CL_FHL`)
Escolhemos `CL_FHL`, que, segundo o dicionário de dados da base, representa o **número de filhos do cliente** (valores de 0 a 4). Não usamos `CL_EC` (código de estado civil) nem os identificadores `CO_ID`, `CL_ID` e `PR_ID`, pois não são medidas numéricas com significado.

**Unidade analisada: o cliente.** `CL_FHL` é uma característica do cliente e se repete em todos os itens que ele compra. Para que clientes com mais compras não pesem mais na média, mantemos **uma linha por cliente** antes de calcular.

```python
# Confirma que cada cliente tem apenas um valor de CL_FHL
qtd_filhos_por_cliente = df_2021.groupby('CL_ID')['CL_FHL'].nunique().max()

print(f'Valores distintos de CL_FHL por cliente (máximo): {qtd_filhos_por_cliente}')

# Uma linha por cliente
df_2021_clientes = df_2021.drop_duplicates(subset='CL_ID')
```

```python
# Estatísticas básicas
estatisticas = df_2021_clientes['CL_FHL'].agg(['count', 'mean', 'min', 'max'])
estatisticas.index = ['Contagem (clientes)', 'Média (filhos)', 'Mínimo', 'Máximo']
print("\n Estatísticas básicas \n")
print(estatisticas.to_frame('CL_FHL').round(2))
```

```python
# Distribuição, para ajudar na interpretação

resumo_filhos = df_2021_clientes['CL_FHL'].value_counts().sort_index().reset_index()

resumo_filhos.columns = ['Nº de Filhos', 'Total Clientes']
print("\n Distribuição \n")
print(resumo_filhos.to_string(index=False))
```

```python
# Contraste: média calculada sobre as linhas (não é a unidade correta)
print('Média sobre os registros (linhas):', round(df_2021['CL_FHL'].mean(), 2))
```

**Interpretação (unidade: cliente):**

*Média de 1,14 filhos por cliente:* em média, cada um dos 995 clientes com compras em 2021 tem pouco mais de um filho.

*Mínimo 0 e máximo 4, com 522 clientes (52,5%) sem filhos:* mais da metade dos clientes não tem filhos, por isso a mediana é 0 e a média fica puxada para cima pelos clientes com 3 ou 4 filhos. Se a média fosse calculada sobre os registros (itens), o resultado seria 1,16, pois clientes que compram mais itens pesariam mais; por isso a unidade correta é o cliente.

### 4.3. Agrupamento 1: registros por categoria de produto
Contamos os registros (itens adquiridos) de 2021 em cada categoria (`PR_CAT`). O resultado mede **quantos itens foram comprados por categoria**; não representa faturamento, nem número de compras ou de clientes.

```python
# Conta os registros de cada categoria
contagem = df_2021.groupby('PR_CAT').size()

# Ordena do maior para o menor
contagem = contagem.sort_values(ascending=False)

# Monta a tabela com a coluna de registros
registros_categoria = contagem.to_frame('Registros')

# Calcula o percentual de cada categoria
total = registros_categoria['Registros'].sum()
registros_categoria['% do total'] = (registros_categoria['Registros'] / total * 100).round(1)

print(registros_categoria)
```

**Interpretação:** `ALIMENTOS` concentra 52,5% dos itens vendidos em 2021 (113.924 registros), seguida de `HIGIENE` (18,8%) e `LIMPEZA` (17,4%). A categoria `NAO INFORMADO` (963 registros, 0,4%) vem dos valores `#N/D` tratados na Sprint 2 e tem peso pequeno.

### 4.4. Agrupamento 2: registros, compras e clientes por segmento
Para cada segmento econômico do cliente (`CL_SEG`, classes A, B e C) calculamos três medidas diferentes: **registros** (itens), **compras** (`CO_ID` distintos) e **clientes** (`CL_ID` distintos). Também calculamos quantas compras cada cliente fez, em média, e quantos itens há por compra.

```python
segmentos = df_2021.groupby('CL_SEG').agg(
    registros = ('CO_ID', 'size'),
    compras = ('CO_ID', 'nunique'),
    clientes = ('CL_ID', 'nunique'))

segmentos['compras_por_cliente'] = (segmentos['compras'] / segmentos['clientes']).round(2)

segmentos['itens_por_compra'] = (segmentos['registros'] / segmentos['compras']).round(1)

print(segmentos)
```

**Interpretação:** A classe `B` concentra a maior parte da atividade (138.313 registros, 3.510 compras e 642 clientes), seguida por `C` e `A`. No entanto, as compras por cliente (entre 5,47 e 5,67) e os itens por compra (cerca de 39) são muito parecidos entre os segmentos. Ou seja, `B` aparece na frente porque **tem mais clientes**, não porque cada cliente compre mais.
